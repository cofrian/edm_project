# 05 · Optimización Valenbisi (Modo A) — PuLP

Selección de **N** ubicaciones que maximizan `α·tráfico + β·población + γ·déficit`.

- Variable: `x_i ∈ {0,1}`.
- Objetivo: `max Σ score_i · x_i`.
- Restricción: `Σ x_i = N`.
- Solver: PuLP (CBC).

La presión de tráfico de cada candidato proviene de la predicción CatBoost (narrativa única).

```python
import sys, os
sys.path.insert(0, os.path.abspath(os.path.join('..', 'backend')))
os.environ['DATA_DIR'] = os.path.abspath(os.path.join('..', 'backend', 'data', 'processed'))
from src.schemas import ValenbisiRequest
from src.optimize_valenbisi import optimize

res = optimize(ValenbisiRequest(n=10, alpha_trafico=1.5, beta_poblacion=1.0, gamma_deficit=1.0))
print('seleccionados:', res.n_selected, '| score total:', res.total_score, '| restriccion:', res.constraint)
for s in res.selected[:10]:
    print(s.candidate_id, 'zona', s.zona, 'score', s.score, 'coste', s.cost)
```
