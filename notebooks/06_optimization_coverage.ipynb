# 06 · Optimización de cobertura urbana (Modo B) — PuLP

Selección de actuaciones bajo **presupuesto** que maximizan
`α·población_cubierta + β·presión_tráfico + γ·déficit_cobertura`.

- Variable: `x_i ∈ {0,1}`.
- Objetivo: `max Σ score_i · x_i`.
- Restricción: `Σ coste_i · x_i ≤ presupuesto`.
- Solver: PuLP (CBC).

Adaptación de la formulación de cobertura '15-min city' del notebook original a movilidad/presión urbana.

```python
import sys, os
sys.path.insert(0, os.path.abspath(os.path.join('..', 'backend')))
os.environ['DATA_DIR'] = os.path.abspath(os.path.join('..', 'backend', 'data', 'processed'))
from src.schemas import CoverageRequest
from src.optimize_coverage import optimize

res = optimize(CoverageRequest(presupuesto=100, alpha_poblacion=1.0, beta_trafico=1.5, gamma_deficit=1.0))
print('seleccionados:', res.n_selected, '| coste total:', res.total_cost, '| restriccion:', res.constraint)
assert res.total_cost <= 100 + 1e-6, 'el coste no puede superar el presupuesto'
for s in res.selected[:10]:
    print(s.candidate_id, 'zona', s.zona, 'score', s.score, 'coste', s.cost)
```
